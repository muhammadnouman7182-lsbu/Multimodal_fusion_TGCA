```python
import os
import torch
import matplotlib.pyplot as plt
import numpy as np
import argparse # Needed for the namespace object saved in your pt files

# --- 1. SET YOUR PATHS HERE ---
# Based on your logs, this is the default save path structure:
checkpoints_base_dir = 'checkpoints/TCGA_GBMLGG/unimodal_path/omic'
model_name = 'omic'
num_splits_to_plot = 15 # Or however many have finished so far (e.g., 9 or 15)

# --- 2. EXTRACT METRICS FROM CHECKPOINTS ---
splits = []
train_cindex = []
test_cindex = []

for k in range(1, num_splits_to_plot + 1):
    checkpoint_path = os.path.join(checkpoints_base_dir, f'{model_name}_{k}.pt')

    if not os.path.exists(checkpoint_path):
        print(f"Waiting for split {k} to finish... Skipping.")
        continue

    try:
        # Load the checkpoint. We set weights_only=False to allow PyTorch to load
        # the argparse.Namespace object containing your training options.
        checkpoint = torch.load(checkpoint_path, map_location=torch.device('cpu'), weights_only=False)

        # Extract the final C-Index values from the metric_logger
        final_train_c = checkpoint['metrics']['train']['cindex'][-1]
        final_test_c = checkpoint['metrics']['test']['cindex'][-1]

        splits.append(k)
        train_cindex.append(final_train_c)
        test_cindex.append(final_test_c)

        print(f"Loaded Split {k} => Train: {final_train_c:.4f} | Test: {final_test_c:.4f}")

    except Exception as e:
        print(f"Error loading {checkpoint_path}: {e}")

# --- 3. GENERATE THE PLOT ---
if len(splits) > 0:
    plt.figure(figsize=(10, 6), dpi=120)

    # Plot lines
    plt.plot(splits, train_cindex, marker='o', linestyle='-', color='dodgerblue', label='Train C-Index', linewidth=2.5, markersize=8)
    plt.plot(splits, test_cindex, marker='s', linestyle='-', color='darkorange', label='Test C-Index', linewidth=2.5, markersize=8)

    # Add shading to visualize the gap
    plt.fill_between(splits, train_cindex, test_cindex, color='gray', alpha=0.1)

    # Formatting
    plt.title('PathomicFusion : C-Index Performance per Split', fontsize=16, pad=15, fontweight='bold')
    plt.xlabel('Cross-Validation Split', fontsize=14)
    plt.ylabel('Concordance Index (C-Index)', fontsize=14)

    # Adjust Y-axis dynamically based on the max/min
    min_y = min(min(test_cindex), min(train_cindex)) - 0.05
    max_y = max(max(test_cindex), max(train_cindex)) + 0.05
    plt.ylim(min_y, max_y)

    plt.xticks(splits)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(fontsize=12, loc='lower right')

    # Add average lines and text labels
    mean_train = np.mean(train_cindex)
    mean_test = np.mean(test_cindex)

    plt.axhline(mean_train, color='dodgerblue', linestyle='--', alpha=0.5)
    plt.axhline(mean_test, color='darkorange', linestyle='--', alpha=0.5)

    last_x = splits[-1] + 0.2
    plt.text(last_x, mean_test, f'Avg Test: {mean_test:.3f}', va='center', color='darkorange', fontsize=11, fontweight='bold')
    plt.text(last_x, mean_train, f'Avg Train: {mean_train:.3f}', va='center', color='dodgerblue', fontsize=11, fontweight='bold')

    plt.tight_layout()

    # Save and show
    output_img = os.path.join(checkpoints_base_dir, 'cindex_checkpoint_performance.png')
    # If the directory doesn't exist for some reason, just save to current directory
    if not os.path.exists(checkpoints_base_dir):
        output_img = 'cindex_checkpoint_performance.png'

    plt.savefig(output_img, dpi=300, bbox_inches='tight')
    print(f"\nPlot saved successfully to: {output_img}")
    plt.show()
else:
    print("No checkpoints were successfully loaded. Please make sure the paths are correct.")
```

```python
import os
import pickle
import numpy as np
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test
import argparse

def plot_km_curves(opt):
    # 1. Path to the prediction file
    # By default, train_cv.py saves test predictions as:
    # {checkpoints_dir}/{exp_name}/{model_name}/{model_name}_{split}_pred_test.pkl

    # We will try to find the prediction file for Split 1 as a sample
    split_idx = 1
    pred_path = os.path.join(opt.checkpoints_dir, opt.exp_name, opt.model_name, f"{opt.model_name}_{split_idx}_pred_test.pkl")

    if not os.path.exists(pred_path):
        # Fallback for earlier versions or patch-based runs
        pred_path = os.path.join(opt.checkpoints_dir, opt.exp_name, opt.model_name, f"{opt.model_name}_{split_idx}_patch_pred_test.pkl")

    if not os.path.exists(pred_path):
        print(f"Error: Could not find prediction file at {pred_path}")
        print("Please check your --checkpoints_dir and --exp_name.")
        return

    print(f"Loading predictions from: {pred_path}")
    with open(pred_path, 'rb') as f:
        pred_data = pickle.load(f)

    # Structure of pred_test: [hazards, surv_times, censors, probs, ground_truth]
    hazards = pred_data[0]
    surv_times = pred_data[1]
    censors = pred_data[2]

    # 2. Split into High vs Low Risk groups based on median hazard
    median_risk = np.median(hazards)
    high_risk_idx = hazards > median_risk
    low_risk_idx = hazards <= median_risk

    # 3. Initialize Kaplan-Meier Fitter
    kmf_high = KaplanMeierFitter()
    kmf_low = KaplanMeierFitter()

    plt.figure(figsize=(10, 7))

    # 4. Fit and Plot
    kmf_high.fit(surv_times[high_risk_idx], event_observed=censors[high_risk_idx], label=f'High Risk (n={sum(high_risk_idx)})')
    ax = kmf_high.plot_survival_function(color='red', linestyle='--')

    kmf_low.fit(surv_times[low_risk_idx], event_observed=censors[low_risk_idx], label=f'Low Risk (n={sum(low_risk_idx)})')
    kmf_low.plot_survival_function(ax=ax, color='blue')

    # 5. Log-Rank Test for statistical significance
    results = logrank_test(surv_times[high_risk_idx], surv_times[low_risk_idx],
                           event_observed_A=censors[high_risk_idx], event_observed_B=censors[low_risk_idx])

    p_value = results.p_value

    # Formatting the plot
    plt.title(f'Kaplan-Meier Survival Curves (Split {split_idx})\nLog-rank p-value: {p_value:.2e}', fontsize=15)
    plt.xlabel('Time (Months)', fontsize=12)
    plt.ylabel('Survival Probability', fontsize=12)
    plt.grid(True, which="both", ls="-", alpha=0.5)
    plt.legend()

    # Save and show
    save_fig_path = os.path.join(opt.checkpoints_dir, opt.exp_name, opt.model_name, f'km_curve_split_{split_idx}.png')
    plt.savefig(save_fig_path)
    print(f"Plot saved to: {save_fig_path}")
    plt.show()

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument('--checkpoints_dir', type=str, default='./checkpoints/TCGA_GBMLGG', help='models are saved here')
    parser.add_argument('--exp_name', type=str, default='unimodal_path', help='name of the experiment')
    parser.add_argument('--model_name', type=str, default='omic', help='model_name subfolder')

    opt = parser.parse_known_args()[0]
    plot_km_curves(opt)
```
